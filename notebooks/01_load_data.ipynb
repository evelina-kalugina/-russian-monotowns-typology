## Load all indicators

```python
import glob

# load every yearly file of population and migration into one table each
raw = {}
for name in ["population", "migration"]:
    paths = sorted(glob.glob(data + f"raw/{name}_data_*.csv"))
    df = pd.concat([pd.read_csv(p, sep=";", dtype=str) for p in paths], ignore_index=True)
    df["indicator_value"] = pd.to_numeric(df["indicator_value"], errors="coerce")
    df["year"] = df["year"].astype(int)
    raw[name] = df
    print(f"--- {name} ({len(paths)} files) ---")
    print("shape:", df.shape)
    print("years:", sorted(df["year"].unique()))
    print()
```

```python
m = raw["migration"]
print(list(m.columns))
print()
for col in m.columns:
    if m[col].nunique() < 50:
        print(col, "->", m[col].unique().tolist())
```

```python
for name in ["wages", "employees", "profitable"]:
    paths = sorted(glob.glob(data + f"raw/{name}_data_*.csv"))
    df = pd.concat([pd.read_csv(p, sep=";", dtype=str) for p in paths], ignore_index=True)
    df["indicator_value"] = pd.to_numeric(df["indicator_value"], errors="coerce")
    df["year"] = df["year"].astype(int)
    raw[name] = df
    print(f"--- {name} ({len(paths)} files) ---")
    print("shape:", df.shape)
    print("years:", sorted(df["year"].unique()))
    print("columns:", list(df.columns))
    print()
```

```python
m = raw["migration"]
mask = (m["migr"].str.startswith("Миграция")) & (m["grup_2"] == "Всего") & (m["vozr"] == "Всего")
raw["migration"] = m[mask].copy()
del m
print(raw["migration"].shape)
```

```python
for name in ["wages", "employees", "profitable"]:
    print(f"--- {name} ---")
    print(raw[name]["okved2"].value_counts().head(25))
    print()
```

```python
TOTAL = "Всего по обследуемым видам экономической деятельности"

clean = {
    "population": raw["population"][raw["population"]["mest"] == "Все население"],
    "migration":  raw["migration"],
    "wages":      raw["wages"][raw["wages"]["okved2"] == TOTAL],
    "employees":  raw["employees"][raw["employees"]["okved2"] == TOTAL],
    "profitable": raw["profitable"][raw["profitable"]["okved2"] == TOTAL],
}

# jobs in mining and manufacturing, industry-dependence
emp = raw["employees"]
industry = emp[emp["okved2"].str.contains("Добыча|Обрабатывающие", na=False)]

for name, df in clean.items():
    dups = df.duplicated(subset=["oktmo", "year"]).sum()
    print(f"{name}: {df.shape[0]} rows, {df['oktmo'].nunique()} municipalities, duplicates: {dups}")
print(f"industry: {industry.shape[0]} rows")
```

```python
def what_differs(df):
    d = df[df.duplicated(subset=["oktmo", "year"], keep=False)]
    n = d.groupby(["oktmo", "year"]).nunique()
    return n.columns[(n > 1).any()].tolist()

for name in ["population", "wages", "employees"]:
    print(name, "-> columns that differ:", what_differs(clean[name]))

print()
print(clean["wages"]["indicator_period"].value_counts())
print()
print(clean["employees"].groupby("year")["indicator_period"].value_counts())
print()
w = clean["wages"]
print(w[w["municipality"].str.contains("Сланцев", na=False)][["municipality", "mun_type", "indicator_period", "indicator_value"]])
```

```python
#keep full-year values only
for name in ["wages", "employees"]:
    clean[name] = clean[name][clean[name]["indicator_period"] == "Январь-декабрь"]

#show the population duplicate
p = clean["population"]
print(p[p.duplicated(subset=["oktmo", "year"], keep=False)][
    ["region_name", "municipality", "mun_type", "oktmo", "year", "indicator_value"]])
print()

#check duplicates again
for name, df in clean.items():
    dups = df.duplicated(subset=["oktmo", "year"]).sum()
    print(f"{name}: {df.shape[0]} rows, {df['oktmo'].nunique()} municipalities, duplicates: {dups}")
```

```python
#removing duplicate population rows (same oktmo and year):
#Makushinsky district 2017 (values differ by 1 person), Novokuznetsk 2022 (same value listed as okrug and intracity district)
clean["population"] = clean["population"].drop_duplicates(subset=["oktmo", "year"], keep="first")
print(clean["population"].duplicated(subset=["oktmo", "year"]).sum())
```

## Save clean tables

```python
import os
os.makedirs(data + "processed", exist_ok=True)

keep = ["region_name", "mun_district", "municipality", "mun_type",
        "oktmo", "oktmo_stable", "year", "indicator_value"]

for name, df in clean.items():
    df[keep].to_csv(data + f"processed/{name}.csv", index=False, encoding="utf-8-sig")
# industry: keep only full-year values, as for wages and employees
industry.loc[industry["indicator_period"] == "Январь-декабрь", keep + ["okved2"]].to_csv(
    data + "processed/industry.csv", index=False, encoding="utf-8-sig"
)
print("saved")
```
