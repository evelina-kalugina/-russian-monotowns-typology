```python
import pandas as pd

data = "../data/"
```

```python
#Dataset 1
mono = pd.read_csv(data + "monotowns_2026.csv", encoding="utf-8-sig")
print(mono.shape)
mono.head()
```

```python
#Assigning the category to the monotowns (they are split into three categories)
mono["category"] = pd.cut(mono["number"], bins=[0, 52, 201, 264], labels=[1, 2, 3])
mono["category"].value_counts()
```
